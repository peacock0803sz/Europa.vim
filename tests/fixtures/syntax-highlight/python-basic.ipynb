```python
def add(a, b):
    """Add two numbers."""
    return a + b

# Call the function
result = add(40, 2)
print(result)
```

```python
class Greeter:
    """A simple greeter class."""

    def __init__(self, name: str) -> None:
        self.name = name

    def greet(self) -> str:
        return f'Hello, {self.name}!'
```

```python
greeter = Greeter('Europa')
greeter.greet()
```
